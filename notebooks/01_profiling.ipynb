```python
dup_rows = df.duplicated().sum()
rows_with_missing = df.isna().any(axis=1).sum()
bad_rows = dup_rows + rows_with_missing
error_rate = bad_rows / len(df) * 100

print("Duplicate rows:", dup_rows)
print("Rows with at least one missing value:", rows_with_missing)
print(f"Starting error rate: {error_rate:.1f}%")
```
