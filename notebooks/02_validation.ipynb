```python
import pandas as pd

raw = pd.read_csv("../data/raw_orders_messy.csv", dtype=str, keep_default_na=False)
clean = pd.read_csv("../data/orders_clean.csv", dtype=str, keep_default_na=False)

print("raw:", raw.shape)
print("clean:", clean.shape)
```

```python
# Pick 200 random raw rows (random_state makes the pick repeatable)
sample = raw.sample(n=200, random_state=42).copy()

# How many times does each picked ID appear in the raw file?
sample["times_in_raw"] = sample["record_id"].map(raw["record_id"].value_counts())

# Find each order's cleaned version using record_id
merged = sample.merge(clean, on="record_id", how="left", suffixes=("_raw", "_clean"))

# Put raw and clean versions side by side
cols = ["record_id", "times_in_raw"]
for c in ["order_date", "product_name", "batch_number", "quantity",
          "unit_price", "customer_name", "region", "status"]:
    cols += [c + "_raw", c + "_clean"]
cols += ["needs_review"]

out = merged[cols].copy()
out["my_check"] = ""   # an empty column for you to fill in

out.to_excel("../data/validation_sample.xlsx", index=False)

print("Sheet size:", out.shape)
print("Rows with no match in the clean data:", out["order_date_clean"].isna().sum())
```
